```python

import sys
import os
from dotenv import load_dotenv
from openai import OpenAI 

load_dotenv()  

GEMINI_API_URL = "https://generativelanguage.googleapis.com/v1beta/openai"
GEMINI_API_KEY = os.getenv("GEMINI_API_KEY")
LLM_ID = "gemini-2.0-flash-lite"

client = OpenAI(
    base_url=GEMINI_API_URL,
    api_key=GEMINI_API_KEY
)

def ai_chat(messages: list):
    # openai api package를 통해 gemini api를 호출합니다.
    # print(f"GEMINI API 호출, MODEL={LLM_ID}")
    response = client.chat.completions.create(
        model=LLM_ID,
        messages=messages,
    )
    return response

ai_chat([
  {"role": "system", "content": "너는 유치원생이야. 유치원생처럼 답변해 줘"},
  {"role": "user", "content": "참새"},
  {"role": "assistant", "content": "짹짹"}, # one-shot prompting
  {"role": "user", "content": "말"},
    {"role": "assistant", "content": "히이잉"}, # one-shot prompting
  {"role": "user", "content": "개구리"},
    {"role": "assistant", "content": "개굴개굴"}, # one-shot prompting
  {"role": "user", "content": "고추"},
]).choices[0].message.content
```

```python
def get_ai_response(prompt: str) -> str:

    response = client.chat.completions.create(
        model=LLM_ID,
        temperature=0.9,
        messages=messages,
    )
    return response.choices[0].message.content
  
messages = [
      {"role": "system", "content": "너는 사용자를 도와주는 상담사야."},
  ]

while True:
    user_input = input("사용자: ")
    if user_input.lower() in ["종료", "exit", "quit"]:
        print("대화를 종료합니다.")
        break

    messages.append({"role": "user", "content": user_input})
    ai_response = get_ai_response(messages)
    messages.append({"role": "assistant", "content": ai_response})
    print("AI:", ai_response)
```

```python
# pdf without header and footer


import pymupdf
import os

pdf_path = "data/interview.pdf"
doc = pymupdf.open(pdf_path)

header_height = 80  # Adjust this value based on your PDF's header height
footer_height = 80  # Adjust this value based on your PDF's footer height

full_text = ""

for page in doc :
    rect = page.rect
    
    header = page.get_text(clip=(0,0,rect.width, header_height)) # 	clip 파라미터: (x0, y0, x1, y1) 좌표로 지정한 영역만 텍스트를 추출
    footer = page.get_text(clip=(0,rect.height - footer_height,rect.width, rect.height))
    text = page.get_text(clip=(0,header_height,rect.width, rect.height - footer_height))
    
    full_text += text + "\n--------------------------------\n"
    
    # 파일명만 추출
pdf_file_name = os.path.basename(pdf_path)
pdf_file_name = os.path.splitext(pdf_file_name)[0] # 확장자 제거

txt_file_path = f"data/{pdf_file_name}_no_header_footer.txt"
with open(txt_file_path, "w", encoding="utf-8") as txt_file:
    txt_file.write(full_text)
```

```python
# 논문 요약해주는 AI 연구원 완성하기


import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()

GEMINI_API_URL = "https://generativelanguage.googleapis.com/v1beta/openai"
GEMINI_API_KEY = os.getenv("GEMINI_API_KEY")
LLM_ID = "gemini-2.0-flash-lite"

def summarize_txt(file_path: str):
    client = OpenAI(
        base_url=GEMINI_API_URL,
        api_key=GEMINI_API_KEY
    )   
    
    with open(file_path, "r", encoding="utf-8") as file:
        txt = file.read()
        
    system_prompt = """
    너는 다음 글을 요약하는 봇이다. 아래 글을 읽고, 저자의 문제 인식과 주장을 파악하고, 주요내용을 요약하라.
    
    작성해야 하는 포맷은 다음과 같다.
    
    # 제목
    
    ## 저자의 문제인식 및 주장 ( 15문장 이내)
    
    ## 저자 소개
    
    ========================= 이하 텍스트 =========================
    
    {txt}
    """
    print(system_prompt)
    print("========================================")
    
    response = client.chat.completions.create(
        model=LLM_ID,
        temperature=0.1,
        messages=[
            {"role": "system", "content": system_prompt},
        ])
    
    return response.choices[0].message.content
  
if __name__ == "__main__":
    file_path = "data/interview_no_header_footer.txt"  # 요약할 텍스트 파일 경로
    # summarize_txt 내부가 시스템 프롬프트만 보내므로, 여기서 사용자 메시지에 본문을 넣어 호출
    with open(file_path, "r", encoding="utf-8") as f:
        txt = f.read()

    system_prompt = (
        "너는 다음 글을 요약하는 봇이다. 아래 글을 읽고, 저자의 문제 인식과 주장을 파악하고, 주요내용을 요약하라.\n\n"
        "# 제목\n\n"
        "## 저자의 문제인식 및 주장 ( 15문장 이내)\n\n"
        "## 저자 소개"
    )

    response = client.chat.completions.create(
        model=LLM_ID,
        temperature=0.1,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": txt},
        ],
    )
    summary = response.choices[0].message.content
    print("===== 요약 결과 =====")
    print(summary)
    
    with open("data/summary.txt", "w", encoding="utf-8") as summary_file:
        summary_file.write(summary)
    
```

```python
# 이미지 활용해 퀴즈 만들기

from glob import glob # for 문으로 여러 파일 처리할 때 유용
from openai import OpenAI
from dotenv import load_dotenv
import os
import base64
load_dotenv()

GEMINI_API_URL = "https://generativelanguage.googleapis.com/v1beta/openai"
GEMINI_API_KEY = os.getenv("GEMINI_API_KEY")
LLM_ID = "gemini-2.0-flash-lite"  

def encode_image_to_base64(image_path: str) -> str:
    with open(image_path, "rb") as image_file:
        encoded_string = base64.b64encode(image_file.read()).decode('utf-8')
    return encoded_string
  
def image_quiz(image_path: str):
    quiz_prompt = """
    제공한 이미지를 바탕으로, 다음과 같은 형식으로 퀴즈를 만들어 주세요.
    정답은 (1) ~ (4) 중 하나만 해당하도록 출제하세요.
    아래는 예시입니다.
    ------- 예시 --------
    
    Q: 다음 이미지에 대한 설명 중 옳지 않은 것은 무엇인가요?
    - (1) 베이커리에서 사람들이 빵을 사는 모습이 담겨 있습니다.
    - (2) 맨앞에 서 있는 여성은 빨간색 코트를 입고 있습니다.
    - (3) 배경에는 다양한 빵들이 진열되어 있습니다.
    - (4) 이미지 속 인물들은 모두 모자를 쓰고 있습니다.
    
    (정답): (4) 이미지 속 인물들은 모두 모자를 쓰고 있습니다. 몇몇 인물들은 모자를 쓰지 않았습니다.
    (주의: 정답은 반드시 (1) ~ (4) 중 하나만 해당하도록 작성해주세요.)
    """
    
    messages = [
        {"role": "user", "content": [
            {"type": "text", "text": quiz_prompt},
            {"type": "image_url", "image_url": {"url": f"data:image/png;base64,{encode_image_to_base64(image_path)}"}}]}
        ]
    
    response = client.chat.completions.create(
        model=LLM_ID,
        messages=messages)
    return response.choices[0].message.content
  

q = image_quiz("data/test.png")
print(q)
  
```

```python
# 이미지 활용해 영어퀴즈 만들기

from glob import glob  # for 문으로 여러 파일 처리할 때 유용
from openai import OpenAI
from dotenv import load_dotenv
import os
import base64
import json

load_dotenv()

GEMINI_API_URL = "https://generativelanguage.googleapis.com/v1beta/openai"
GEMINI_API_KEY = os.getenv("GEMINI_API_KEY")
LLM_ID = "gemini-2.0-flash-lite"

def encode_image_to_base64(image_path: str) -> str:
    with open(image_path, "rb") as image_file:
        encoded_string = base64.b64encode(image_file.read()).decode('utf-8')
    return encoded_string

def image_quiz(image_path: str, n_trial=0, max_trial=3):
  
    if n_trial >= max_trial:  # 최대시도 횟수에 도달하면 포기
        raise Exception("Max trial reached for encoding image.")

    base64_image =encode_image_to_base64(image_path)

    quiz_prompt = """
    제공한 이미지를 바탕으로, 다음과 같은 형식으로 퀴즈를 만들어 주세요.
    정답은 (1) ~ (4) 중 하나만 해당하도록 출제하세요.
    토익 리스닝 문제 스타일로 문제를 만들어 주세요.
    아래는 예시입니다.
    ------- 예시 --------
    
    Q: 다음 이미지에 대한 설명 중 옳지 않은 것은 무엇인가요?
    - (1) 베이커리에서 사람들이 빵을 사는 모습이 담겨 있습니다.
    - (2) 맨앞에 서 있는 여성은 빨간색 코트를 입고 있습니다.
    - (3) 배경에는 다양한 빵들이 진열되어 있습니다.
    - (4) 이미지 속 인물들은 모두 모자를 쓰고 있습니다.
    
    Listening: Which of the following statements about the image is NOT correct?
    - (1) The image shows people buying bread at a bakery.
    - (2) The woman at the front is wearing a red coat.
    - (3) Various breads are displayed in the background.
    - (4) All the people in the image are wearing hats.
    
    (정답): (4) 이미지 속 인물들은 모두 모자를 쓰고 있습니다. 몇몇 인물들은 모자를 쓰지 않았습니다.
    (주의: 정답은 반드시 (1) ~ (4) 중 하나만 해당하도록 작성해주세요.)
    """

    messages = [
        {
            "role": "user",
            "content": [
                {"type": "text", "text": quiz_prompt},
                {
                    "type": "image_url",
                    "image_url": {
                        "url": f"data:image/png;base64,{encode_image_to_base64(image_path)}"
                    },
                },
            ],
        }
    ]
    try:
        response = client.chat.completions.create(model=LLM_ID, messages=messages)
    except Exception as e:
        print(f"Error: {e}")
    content = response.choices[0].message.content


    if "Listening:" in content:
          return content, True
    else:
          return image_quiz(image_path, n_trial + 1)
      
txt = ''
no = 1
eng_dict = []

for g in glob("data/*.png"):
    
    q, is_succeed = image_quiz(g)
    if not is_succeed:
        print(f"Skipping file {g} due to repeated failures.")
        continue

    divider = f'## 문제 {no} ##\n'
    print(divider)
    
    txt += divider
    filename = os.path.basename(g)
    txt += f'![문제 이미지]({filename})\n\n'
    
    # 문제추가
    print(q)
    txt += q + '\n\n------------------------\n\n'
    
    with open("data/image_quiz.md", "w", encoding="utf-8") as f:
        f.write(txt)
    
   
    # 영어문제만 출력
    
    eng = q.split("Listening:")[1].split("(정답):")[0].strip()
    
    eng_dict.append({
        "no": no,
        "eng": eng,
        "img": filename
    })
    
    with open("data/image_quiz_eng.json", "w", encoding="utf-8") as f:
        json.dump(eng_dict, f, ensure_ascii=False, indent=4)
        
    no += 1
    
    


```
