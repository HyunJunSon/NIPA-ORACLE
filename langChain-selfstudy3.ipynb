```python
from typing import Dict, TypedDict

sample_dict: Dict[str, int] = {
  "name" : "테디",
  "age" : 5,
  "job" : "developer"
}

class Person(TypedDict):
    name: str
    age: int # 정수형으로 명시
    job: str
    
type_dict: Person = {
  "name" : "sally",
  "age" : 5,
  "job" : "designer"
}

sample_dict['age'] = 35
sample_dict['new_filed'] = "added value"

sample_dict # 아무 문제 없음
```

```python
# !mypy langChain-selfstudy3.ipynb    
type_dict['age'] = "thirty five"  # 타입 오류 발생
type_dict['new_field'] = "added value"  # 타입 오류 발생
type_dict
```
