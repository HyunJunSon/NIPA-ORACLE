### 01. Flask

```python
from flask import Flask, render_template, request, redirect, g
import psycopg2
import psycopg2.extras
from dotenv import load_dotenv
import os


app = Flask(__name__)
load_dotenv()

connection_string = f"postgresql://{os.getenv('DB_USER')}:{os.getenv('DB_PASS')}@{os.getenv('DB_HOST')}:{os.getenv('DB_PORT')}/{os.getenv('DB_NAME')}"
def get_connection():
    if "conn" not in g:
        g.conn = psycopg2.connect(connection_string)
        g.cursor = g.conn.cursor(cursor_factory=psycopg2.extras.DictCursor)
    return g.conn, g.cursor

@app.route("/home")
def home():
  return "This is home"

@app.route("/hello/<name>")
def hello(name):
  return render_template("user.html", myname=name)

@app.route("/")
def index():
  conn,cursor = get_connection()
  cursor.execute("SELECT * FROM test_post")
  contents = cursor.fetchall()
  conn.close()
  return render_template("index.html", contents=contents)

@app.route("/add", methods=("GET","POST"))
def add():
  if request.method == 'POST' :
    title = request.form['title']
    content = request.form['content']
    conn,cursor = get_connection()
    cursor.execute('INSERT INTO test_post (title,content) values (%s, %s)', (title,content))
    conn.commit()
    conn.close()
    return redirect("/")
  
  elif request.method == 'GET' :
    return render_template("add.html")
  return render_template("add.html")



if __name__ =="__main__" :
  app.run( host="0.0.0.0")
```
